# Model Evaluation: Latihan Interaktif

Notebook ini berisi latihan hands-on untuk evaluation metrics dan cross-validation.

Lihat `README.md` untuk teori lengkap.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.linear_model import LogisticRegression

from implementation import (
    train_test_split, confusion_matrix, accuracy_score,
    precision_score, recall_score, f1_score,
    roc_curve, roc_auc_score, cross_val_score,
    classification_report, plot_confusion_matrix, plot_roc_curve_viz
)

np.random.seed(42)
print("✓ Setup complete!")
```

## Latihan 1: Train/Test Split & Basic Metrics

```python
# Generate data
X, y = make_classification(n_samples=200, n_features=10, random_state=42)

# Split
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Train model
model = LogisticRegression(random_state=42)
model.fit(X_train, y_train)
y_pred = model.predict(X_test)

# Evaluate
print(f"Accuracy:  {accuracy_score(y_test, y_pred):.4f}")
print(f"Precision: {precision_score(y_test, y_pred):.4f}")
print(f"Recall:    {recall_score(y_test, y_pred):.4f}")
print(f"F1 Score:  {f1_score(y_test, y_pred):.4f}")
```

## Latihan 2: Confusion Matrix

```python
cm = confusion_matrix(y_test, y_pred)
print("Confusion Matrix:")
print(cm)

# Plot
fig = plot_confusion_matrix(cm, class_names=['Class 0', 'Class 1'])
plt.show()
```

## Latihan 3: ROC-AUC

```python
# Get probabilities
y_proba = model.predict_proba(X_test)[:, 1]

# Compute ROC curve
fpr, tpr, thresholds = roc_curve(y_test, y_proba)
auc = roc_auc_score(y_test, y_proba)

print(f"ROC-AUC: {auc:.4f}")

# Plot
fig = plot_roc_curve_viz(fpr, tpr, auc)
plt.show()
```

## Latihan 4: Cross-Validation

```python
# 5-fold CV
cv_scores = cross_val_score(model, X, y, cv=5, scoring='accuracy', random_state=42)

print(f"CV Scores: {cv_scores}")
print(f"Mean: {cv_scores.mean():.4f} (+/- {cv_scores.std():.4f})")
```

## Latihan 5: Classification Report

```python
report = classification_report(y_test, y_pred)
print(report)
```

## Latihan 6: Multiclass Metrics

```python
# Generate multiclass data
X_mc, y_mc = make_classification(n_samples=300, n_features=10, n_classes=3, n_informative=8, random_state=42)
X_train_mc, X_test_mc, y_train_mc, y_test_mc = train_test_split(X_mc, y_mc, test_size=0.3, random_state=42)

model_mc = LogisticRegression(random_state=42, max_iter=200)
model_mc.fit(X_train_mc, y_train_mc)
y_pred_mc = model_mc.predict(X_test_mc)

# Macro, Micro, Weighted
print(f"Macro F1:    {f1_score(y_test_mc, y_pred_mc, average='macro'):.4f}")
print(f"Micro F1:    {f1_score(y_test_mc, y_pred_mc, average='micro'):.4f}")
print(f"Weighted F1: {f1_score(y_test_mc, y_pred_mc, average='weighted'):.4f}")
```
