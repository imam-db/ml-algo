# Neural Networks - Interactive Exercises

Learn Multi-Layer Perceptrons hands-on with SPG outlet stockout prediction.

```python
import numpy as np
import matplotlib.pyplot as plt
from implementation import MLPClassifier, generate_spg_outlet_data, gradient_check

# Set random seed for reproducibility
np.random.seed(42)
```

## Exercise 1: Generate and Explore Data

Generate synthetic SPG outlet data and explore its characteristics.

```python
# Generate data
X, y = generate_spg_outlet_data(n_samples=500, n_features=10, n_classes=3, random_state=42)

print(f"Dataset shape: {X.shape}")
print(f"Number of classes: {len(np.unique(y))}")
print(f"Class distribution: {np.bincount(y)}")

# Split data
split_idx = int(0.8 * len(X))
X_train, X_test = X[:split_idx], X[split_idx:]
y_train, y_test = y[:split_idx], y[split_idx:]
```

## Exercise 2: Train MLP and Visualize Learning

Train a Multi-Layer Perceptron and observe how loss decreases.

```python
# Train MLP
mlp = MLPClassifier(
    hidden_layers=[32, 16],
    activation='relu',
    learning_rate=0.01,
    max_iter=1000,
    batch_size=32,
    random_state=42
)

mlp.fit(X_train, y_train)

# Evaluate
train_acc = mlp.score(X_train, y_train)
test_acc = mlp.score(X_test, y_test)

print(f"Training accuracy: {train_acc:.4f}")
print(f"Test accuracy: {test_acc:.4f}")
```

```python
# Plot learning curve
plt.figure(figsize=(10, 5))
plt.plot(mlp.loss_history_)
plt.xlabel('Epoch (x100)')
plt.ylabel('Loss')
plt.title('Neural Network Learning Curve')
plt.grid(True)
plt.show()
```

## Exercise 3: Verify Gradients

Use numerical gradient checking to verify backpropagation implementation.

```python
# Gradient check on small sample
X_small = X_train[:10]
y_small = y_train[:10]

max_diff, rel_diff = gradient_check(mlp, X_small, y_small, epsilon=1e-5)

print(f"Maximum difference: {max_diff:.2e}")
print(f"Relative difference: {rel_diff:.2e}")
print(f"\nGradient check: {'✓ PASS' if rel_diff < 1e-3 else '✗ FAIL'}")
```

## Exercise 4: Experiment with Architecture

Try different network architectures and observe the effects.

```python
# Compare different architectures
architectures = [
    ([16], "Small (16)"),
    ([32, 16], "Medium (32-16)"),
    ([64, 32, 16], "Large (64-32-16)")
]

results = []

for layers, name in architectures:
    mlp = MLPClassifier(hidden_layers=layers, max_iter=500, random_state=42)
    mlp.fit(X_train, y_train)
    acc = mlp.score(X_test, y_test)
    results.append((name, acc))
    print(f"{name}: Test accuracy = {acc:.4f}")
```

## Exercise 5: Prediction Analysis

Analyze predictions and class probabilities.

```python
# Get predictions and probabilities
predictions = mlp.predict(X_test[:10])
probabilities = mlp.predict_proba(X_test[:10])

print("Sample Predictions:")
print("="*50)
for i in range(10):
    print(f"True: {y_test[i]}, Predicted: {predictions[i]}")
    print(f"  Probabilities: {probabilities[i]}")
    print()
```
