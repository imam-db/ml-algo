# K-Nearest Neighbors: Latihan Interaktif

Notebook ini berisi latihan hands-on untuk memahami KNN.

Lihat `README.md` untuk teori lengkap.

```python
# Import libraries
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification, load_iris
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from implementation import KNNClassifier, KNNRegressor, choose_k_elbow_method, plot_k_selection

np.random.seed(42)
print("✓ Setup complete!")
```

## Latihan 1: KNN Classification

```python
# Load Iris dataset
iris = load_iris()
X, y = iris.data, iris.target

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# IMPORTANT: Scale features!
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train)
X_test_scaled = scaler.transform(X_test)

# Train KNN
knn = KNNClassifier(n_neighbors=5)
knn.fit(X_train_scaled, y_train)

# Evaluate
accuracy = knn.score(X_test_scaled, y_test)
print(f"Accuracy: {accuracy:.4f}")
```

## Latihan 2: Choosing Optimal K

```python
# Find optimal k
k_values, scores = choose_k_elbow_method(
    X_train_scaled, y_train,
    X_test_scaled, y_test,
    k_range=range(1, 21),
    task='classification'
)

# Plot
fig = plot_k_selection(k_values, scores, task='classification')
plt.show()

best_k = k_values[np.argmax(scores)]
print(f"Best k: {best_k}")
```

## Latihan 3: Distance Metrics

```python
# Compare distance metrics
metrics = ['euclidean', 'manhattan', 'minkowski']

for metric in metrics:
    knn = KNNClassifier(n_neighbors=5, metric=metric)
    knn.fit(X_train_scaled, y_train)
    acc = knn.score(X_test_scaled, y_test)
    print(f"{metric}: {acc:.4f}")
```

## Latihan 4: Uniform vs Distance Weighting

```python
# Compare weighting schemes
for weights in ['uniform', 'distance']:
    knn = KNNClassifier(n_neighbors=5, weights=weights)
    knn.fit(X_train_scaled, y_train)
    acc = knn.score(X_test_scaled, y_test)
    print(f"{weights}: {acc:.4f}")
```

## Latihan 5: Feature Scaling Impact

```python
# Without scaling
knn_no_scale = KNNClassifier(n_neighbors=5)
knn_no_scale.fit(X_train, y_train)
acc_no_scale = knn_no_scale.score(X_test, y_test)

# With scaling
knn_scaled = KNNClassifier(n_neighbors=5)
knn_scaled.fit(X_train_scaled, y_train)
acc_scaled = knn_scaled.score(X_test_scaled, y_test)

print(f"Without scaling: {acc_no_scale:.4f}")
print(f"With scaling: {acc_scaled:.4f}")
print(f"Improvement: {(acc_scaled - acc_no_scale)*100:.1f}%")
```
