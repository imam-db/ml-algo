# Naive Bayes: Latihan Interaktif

Notebook ini berisi latihan hands-on untuk Gaussian dan Multinomial Naive Bayes.

Lihat `README.md` untuk teori lengkap.

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

from implementation import GaussianNaiveBayes, MultinomialNaiveBayes

np.random.seed(42)
print("✓ Setup complete!")
```

## Latihan 1: Gaussian Naive Bayes

```python
# Load Iris dataset
iris = load_iris()
X, y = iris.data, iris.target

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# Train Gaussian NB
gnb = GaussianNaiveBayes()
gnb.fit(X_train, y_train)

# Predict
predictions = gnb.predict(X_test)
probabilities = gnb.predict_proba(X_test)
accuracy = gnb.score(X_test, y_test)

print(f"Accuracy: {accuracy:.4f}")
print(f"\nSample predictions:")
for i in range(5):
    print(f"True: {y_test[i]}, Pred: {predictions[i]}, Proba: {probabilities[i]}")
```

## Latihan 2: Multinomial Naive Bayes

```python
# Generate count data (simulating word counts)
np.random.seed(42)
X_counts = []
y_labels = []

for label in range(3):
    for _ in range(30):
        counts = np.random.poisson(lam=2, size=10)
        if label == 0:
            counts[:3] *= 3
        elif label == 1:
            counts[3:6] *= 3
        else:
            counts[6:9] *= 3
        X_counts.append(counts)
        y_labels.append(label)

X_counts = np.array(X_counts)
y_labels = np.array(y_labels)

X_train, X_test, y_train, y_test = train_test_split(X_counts, y_labels, test_size=0.3, random_state=42)

# Train Multinomial NB
mnb = MultinomialNaiveBayes(alpha=1.0)
mnb.fit(X_train, y_train)

accuracy = mnb.score(X_test, y_test)
print(f"Accuracy: {accuracy:.4f}")
```

## Latihan 3: Laplace Smoothing

```python
# Test different alpha values
print("Testing different smoothing parameters:")
for alpha in [0.1, 1.0, 5.0, 10.0]:
    mnb = MultinomialNaiveBayes(alpha=alpha)
    mnb.fit(X_train, y_train)
    acc = mnb.score(X_test, y_test)
    print(f"alpha={alpha:>5.1f}: {acc:.4f}")
```

## Latihan 4: Probability Analysis

```python
# Analyze prediction confidence
probas = gnb.predict_proba(X_test)
confidences = probas.max(axis=1)

print(f"Mean confidence: {confidences.mean():.4f}")
print(f"Min confidence: {confidences.min():.4f}")
print(f"Max confidence: {confidences.max():.4f}")

plt.hist(confidences, bins=20)
plt.xlabel('Confidence')
plt.ylabel('Count')
plt.title('Prediction Confidence Distribution')
plt.show()
```
